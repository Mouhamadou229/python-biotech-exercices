```python
patients_ids=["PO1", "P02", "P03", "P04", "P05", "P06", "P07", "P08", "P09", "P10"]
ratio_tg_sur_hdl_c=[0.46, 1.67, 0.60, 1.71, 0.04, 3.13, 0.23, 1.53, 2.75, 0.86 ]
val_norm=0.86
compteur_faible=0
compteur_modere=0
compteur_severe=0
print("--------CLASSIFICATION DES PATIENTS--------")
for patient_id, tg_sur_hdl_c in zip(patients_ids, ratio_tg_sur_hdl_c):
    if tg_sur_hdl_c < val_norm :
        risk="Risque SMet Faible"
        compteur_faible += 1
    elif tg_sur_hdl_c == val_norm :
        risk="Risque SMet Modéré"
        compteur_modere += 1
    else :
        risk="Risque SMet Sévère"
        compteur_severe += 1
    print(f"Patient {patient_id}: {risk}")

print("------NOMBRE DE PATIENTS SELON LE RISQUE-------")
print(f"Risque SMet Faible: {compteur_faible}")
print(f"Risque SMet Modéré: {compteur_modere}")
print(f"Risque SMet Sévère: {compteur_severe}")
```
